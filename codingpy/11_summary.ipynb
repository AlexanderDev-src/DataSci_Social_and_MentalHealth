# 11 · สรุปผล

จากผู้ตอบ 121 คน ใช้วิเคราะห์ 107 คนที่ผ่านข้อดักความใส่ใจ (หญิง 59 ชาย 48)

| คำถาม | ผล |
|---|---|
| 1 ชั่วโมงใช้งานกับสามมิติ | ไม่สัมพันธ์กับมิติใดเลย r อยู่ระหว่าง −0.15 ถึง −0.09 ทุกค่า p > 0.05 |
| 2 มิติที่เกี่ยวมากที่สุด | ไม่มีมิติใดโดดเด่น ช่วงความเชื่อมั่นคร่อม 0 ทั้งสามมิติ |
| 3 ข่าวเชิงลบกับความวิตกกังวล | ทิศทางเป็นไปตามที่คาด (r = 0.15) แต่ไม่มีนัยสำคัญ (p = 0.134; partial r = 0.17, p = 0.103) |
| 4 ประเภทเนื้อหา | คะแนนของ 4 กลุ่มเนื้อหาไม่ต่างกันในมิติใดเลย |
| 5 โซเชียลมีเดียกับการนอน | ชั่วโมงใช้งานไม่สัมพันธ์กับชั่วโมงนอนหรือคุณภาพการนอน แต่คุณภาพการนอนสัมพันธ์กับภาวะซึมเศร้า (r = −0.28) และความเครียด (r = −0.32) |
| 6 เมื่อควบคุมตัวแปรอื่น | ตัวแปรโซเชียลมีเดียเพิ่ม R² ได้ไม่เกิน 0.045 อย่างไม่มีนัยสำคัญ |
| 7 ความเครียดเมื่อเกรดและการเงินเท่ากัน | b = −0.19 คะแนนต่อชั่วโมง (p = 0.485) ไม่พบความต่าง |

**ข้อค้นพบหลัก**

- เวลาที่ใช้กับโซเชียลมีเดียไม่ได้อธิบายคะแนนภาวะซึมเศร้า ความวิตกกังวล หรือความเครียดของนักศึกษากลุ่มนี้
  ทั้งเมื่อดูตรง ๆ และเมื่อควบคุมการนอน การเรียน และการเงินแล้ว
- ตัวแปรที่เกี่ยวกับคะแนนชัดที่สุดคือ**สภาพการเงิน** กลุ่มเงินไม่พอใช้มีคะแนนภาวะซึมเศร้าเฉลี่ย 16.6 เทียบกับ 6.6 ของกลุ่มสบาย
  (F(2, 104) = 10.91, p < 0.001, η² = 0.17) รองลงมาคือ**คุณภาพการนอน**และ**ความกดดันจากการเรียน**
- ผู้ตอบ 68.2% มีคะแนนความวิตกกังวลเกินเกณฑ์ปกติของ DASS-21 และนอนเฉลี่ย 6.21 ชั่วโมง ต่ำกว่าที่แนะนำ
- DASS-21 ฉบับภาษาไทยมีความเชื่อมั่นยอมรับได้ในกลุ่มนี้ (แอลฟา 0.844, 0.737 และ 0.771)

**ข้อจำกัด**

- เก็บข้อมูลครั้งเดียว (cross-sectional) จึงบอกได้แค่ว่าตัวแปรสัมพันธ์กัน ไม่ใช่ตัวหนึ่งทำให้เกิดอีกตัว
- กลุ่มตัวอย่าง 107 คนจากมหาวิทยาลัยเดียว และทุกตัวแปรเป็นการรายงานตนเอง ช่วงความเชื่อมั่นยังกว้างพอจะมีผลขนาดเล็กซ่อนอยู่
- บางกลุ่มเนื้อหามีผู้ตอบน้อย (กลุ่มข่าว 8 คน) การเทียบกลุ่มจึงมีกำลังต่ำ
- ผลใช้ได้ในระดับกลุ่มเท่านั้น ไม่ใช่การวินิจฉัยรายบุคคล ผู้ที่มีข้อกังวลเรื่องสุขภาพจิตโทรสายด่วนสุขภาพจิต 1323 ได้ตลอด 24 ชั่วโมง

ตัวเลขในข้อความสรุปพิมพ์ตามผลของการรันครั้งนี้ ถ้าข้อมูลเปลี่ยนให้รันใหม่ทุกไฟล์ตามลำดับเลข แล้วแก้ข้อความให้ตรงกับผลลัพธ์ในแต่ละไฟล์
