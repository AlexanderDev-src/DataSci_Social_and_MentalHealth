## คำถามวิจัย

1. เวลาที่ใช้กับโซเชียลมีเดียต่อวัน สัมพันธ์กับคะแนนภาวะซึมเศร้า ความวิตกกังวล และความเครียดอย่างไร และแรงเพียงใด
2. โซเชียลมีเดียสัมพันธ์กับมิติใดมากที่สุด
3. การติดตามข่าวเชิงลบ (การเมือง อาชญากรรม ภัยพิบัติ) สัมพันธ์กับความวิตกกังวลมากกว่ามิติอื่นหรือไม่
4. นักศึกษาที่เสพเนื้อหาต่างประเภทกัน มีคะแนนทั้งสามมิติแตกต่างกันหรือไม่
5. เวลาที่ใช้กับโซเชียลมีเดีย สัมพันธ์กับจำนวนชั่วโมงและคุณภาพการนอนอย่างไร
6. เมื่อควบคุมการนอน การออกกำลังกาย ความกดดันจากการเรียน ผลการเรียน และความเพียงพอทางการเงินแล้ว
   เวลาที่ใช้กับโซเชียลมีเดียยังอธิบายความแตกต่างของคะแนนแต่ละมิติได้อยู่หรือไม่
7. ในกลุ่มนักศึกษาที่มีผลการเรียนและความเพียงพอทางการเงินใกล้เคียงกัน
   ผู้ที่ใช้โซเชียลมีเดียมากกว่ามีคะแนนความเครียดต่างจากผู้ที่ใช้น้อยกว่าหรือไม่

## แหล่งข้อมูล

คณะผู้จัดทำ**เก็บข้อมูลเองด้วยแบบสอบถามออนไลน์** (Google Forms) จากนักศึกษามหาวิทยาลัยขอนแก่น
ระหว่างวันที่ 23 สิงหาคม ถึง 3 กันยายน 2569 ได้คำตอบ 121 ชุด

- แบบวัด DASS-21 ฉบับภาษาไทย 21 ข้อ ข้อละ 0 ถึง 3 คะแนน แบ่งเป็นภาวะซึมเศร้า ความวิตกกังวล และความเครียด มิติละ 7 ข้อ
- คำถามพฤติกรรม: ชั่วโมงใช้โซเชียลมีเดียต่อวัน แพลตฟอร์มหลัก ประเภทเนื้อหาที่เสพ จำนวนวันที่ติดตามข่าวต่อสัปดาห์
- ตัวแปรควบคุม: ชั่วโมงและคุณภาพการนอน จำนวนวันออกกำลังกาย ความกดดันจากการเรียน เกรดเฉลี่ย ความเพียงพอทางการเงิน
- ข้อดักความใส่ใจ 1 ข้อ ผู้ที่ตอบผิดถูกตัดออกจากการวิเคราะห์ตามที่เค้าโครงกำหนด

เดิมตั้งใจใช้ชุดข้อมูลสาธารณะจาก Kaggle แต่ชุดข้อมูลในหัวข้อนี้ที่ตรวจแล้วเป็นข้อมูลสังเคราะห์
(หลักทศนิยมกระจายเท่ากันทั้ง 10 หลัก แพลตฟอร์มผูกกับประเทศแบบตายตัว ตัวแปรคนละเรื่องสัมพันธ์กันสูงถึง 0.80 ถึง 0.95
และไม่มีค่าว่างเลย) จึงเก็บข้อมูลจากผู้ตอบจริงแทน

## วิธีรัน

**เครื่องตัวเอง** ใช้ [uv](https://docs.astral.sh/uv/) จัดการแพ็กเกจ วางไฟล์คำตอบไว้ที่ `codingpy/data/data.csv` แล้วรัน

```bash
uv sync
uv run jupyter lab codingpy/    # เปิดไฟล์ที่ต้องการ แล้วเลือก Run > Run All Cells
```

หรือรันทุกไฟล์ตามลำดับเลขโดยไม่เปิดหน้าจอ

```bash
uv run jupyter nbconvert --to notebook --execute --inplace codingpy/[0-9]*.ipynb
```

เปิดจากโฟลเดอร์ `codingpy/` หรือจากรากของ repo ก็ได้ (เซลล์แรกของทุกไฟล์เปลี่ยนโฟลเดอร์ให้)

**Google Colab** หนึ่ง notebook คือหนึ่ง session จึงต้องอัปโหลด `common.py` เข้า session ก่อนรันทุกครั้ง
`01_data_preparation.ipynb` จะเปิดช่องให้อัปโหลด `data.csv` และ `common.py` จะติดตั้งฟอนต์ภาษาไทยให้
ส่วนไฟล์ `02` เป็นต้นไปต้องมี `data/data_clean.csv` ที่ไฟล์ `01` เขียนไว้อยู่ใน session ด้วย
